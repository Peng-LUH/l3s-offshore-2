```python
import re
from urllib.parse import unquote
import requests

URL = "https://l3s-offshore-plan-2.l3s.uni-hannover.de/l3s-offshore-2/offshore-plan/display-single-vessel-schedule"
payload = response.json()

def filename_from_cd(resp, default="schedule.pdf"):
    cd = resp.headers.get("Content-Disposition", "")
    # RFC 5987 (filename*="utf-8''name.pdf")
    m = re.search(r'filename\*\s*=\s*[^\'"]*\'\'([^;]+)', cd)
    if m:
        return unquote(m.group(1))
    # Plain filename="name.pdf" or filename=name.pdf
    m = re.search(r'filename\s*=\s*"?(?P<name>[^";]+)"?', cd)
    if m:
        return m.group("name")
    return default

# --- send & save ---
with requests.post(URL, json=payload, stream=True, timeout=60) as r:
    try:
        r.raise_for_status()
    except requests.HTTPError:
        # Server sent an error; try to show JSON error message if present
        try:
            print("Error:", r.json())
        except Exception:
            print("Error:", r.text)
        raise

    # Basic content-type sanity check (optional)
    ctype = r.headers.get("Content-Type", "").lower()
    if "pdf" not in ctype:
        # Not a PDF? Print response for debugging.
        print("Unexpected content-type:", ctype)
        print(r.text[:500])
        raise RuntimeError("Expected a PDF response.")

    out_path = filename_from_cd(r)
    with open(out_path, "wb") as f:
        for chunk in r.iter_content(chunk_size=8192):
            if chunk:  # filter out keep-alive chunks
                f.write(chunk)

print("Saved:", out_path)
```
